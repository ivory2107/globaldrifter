<font color="#0077B6" size="5px">**Abstract**</font>
---
<p align="justify"> The California Current is once again in a large marine heatwave while a strong El Niño develops throughout 2026–27. Satellite-based tools such as NOAA's Blobtracker show where the ocean is warm, but not really where it moves or reaches. We used over 1.4 million hourly observations from 341 Global Drifter Program (GDP) drifters (2012–2021), the OISST satellite record, and over 17,000 CalCOFI fish-egg samples to compare two similar-sized events: the damaging 2013–2016 "Blob" and the 2019–2020 heatwave, which stayed offshore. With this, we asked three questions. <strong>RQ1 (detect):</strong> a drifter detector using the OISST 90th-percentile threshold agreed with OISST heatwave days better than a fixed +1 °C rule (F1 0.873 vs 0.819). <strong>RQ2 (track):</strong> sampled warm-condition drifters were closer to shore during the Blob (median 467 vs 800 km), and their mean coastal-entry rate was about eight times as high (1.88 vs 0.23 entries per 100 offshore drifter-days). Model C had the lowest median 24-hour position errors among the three models in the retrospective comparison. <strong>RQ3 (impact):</strong> about four times more spring spawning habitat was in a heatwave during the Blob (44% vs 10% of samples), and sardine spawning moved roughly 8° north while staying in about 12.8 °C water. Adding heatwave exposure improved predictions of where sardine eggs occur in all three model sets we compared (best: gradient boosting, ROC-AUC 0.72); for anchovy, it did not. Finally, we recommend using drifters alongside the Blobtracker as a local check on coastal heatwaves, and treating heatwaves that move to shore as an early signal for sardine spawning grounds. </p>

<font color="#0077B6" size="5px">**1. Introduction**</font>
---
<p align="justify"> This study tests whether surface drifters can add something satellite heatwave maps cannot: where unusually warm water goes, whether it reaches the coast, and what that means for fish. Our industry partner is NOAA's Global Drifter Program (GDP), which keeps around 1,300 satellite-tracked buoys drifting in the world's surface ocean. Each drifter reports its position, velocity, and sea-surface temperature (SST) every hour (Elipot et al., 2016).
Ocean observations are of two kinds. Eulerian instruments, such as moorings and satellites, watch fixed places. Lagrangian instruments, such as drifters, move with the water and report what it experiences along the way. Marine heatwave monitoring today is almost entirely Eulerian: NOAA's California Current Marine Heatwave Tracker ("Blobtracker") maps daily satellite SST anomalies on a fixed grid. A drifter instead follows a parcel of water, so in principle it can show whether a patch of warm water is moving towards the coast.
We study the California Current off the U.S. West Coast, which is currently in a large marine heatwave (NEP25A) as a strong El Niño develops. The GDP hourly record available from Clouddrift ends in October 2022, so it cannot follow NEP25A directly. Instead, we build and test drifter-based methods on two past events fully covered by the data: the 2013–2016 "Blob", which reached the coast and disrupted fisheries, and the 2019–2020 heatwave, which was nearly as large but stayed offshore (Weber et al., 2021). Comparing a damaging event with a less damaging one would be a stronger test than validating on a single event.
 </p>

<font color="#0077B6" size="5px">**2. Background and Motivation**</font>
---
<p align="justify"> The California Current is in a record-breaking heatwave just as a strong El Niño arrives. NOAA has tracked NEP25A as a contiguous feature since May 2025. It reached about 10 million km² on 10 September 2025, the largest heatwave by area in the Northeast Pacific since monitoring began in 1982 (NOAA California Current IEA, 2026). As of late October 2026, NOAA reports that it has expanded and intensified off the southern West Coast as El Niño-driven Kelvin waves arrive. NOAA's Climate Prediction Center issued an El Niño Advisory on 8 October 2026, giving a greater than 83% chance of a strong-to-very strong El Niño through January–March 2027 (NOAA Climate Prediction Center, 2026).

**Why warm water at the coast matters.** The California Current is a wind-driven upwelling system. Equatorward winds push surface water offshore, cold nutrient-rich water rises to replace it, and that nutrient supply feeds one of the world's most productive fisheries. A marine heatwave caps the surface with warm water and weakens this supply. El Niño adds warming and further suppresses upwelling. NOAA notes that the heatwave and El Niño may not simply add together; the larger risk may be how long the coast stays warm (NOAA California Current IEA, 2026).

The Blob formed in the Gulf of Alaska in winter 2013–14 from reduced ocean heat loss (Bond et al., 2015), persisted for several years, and by 2015, stretched from Alaska to Baja California. Its effects included northward shifts of tropical species, closures of commercially important fisheries, a record harmful algal bloom, and mass strandings of marine mammals and seabirds (Cavole et al., 2016). Sardine and anchovy swung between record lows and highs in this period (Thompson et al., 2022).

**What is at stake for fisheries and coastal economies.** The Blob led to a series of U.S. federal fishery disaster declarations along the West Coast, covering Pacific cod, urchin and abalone, salmon, crab and sardine, and over $141 million USD in relief funds to affected fishers, processors, and dealers (Free at al., 2023).

El Niño is already hitting anchovy fisheries elsewhere in the Pacific in 2026. Peru, the world's largest anchovy producer, landed only 2,496 tonnes of anchovy in July 2026, a 99.1% fall from 292,148 tonnes in July 2025, according to national production statistics (INEI, reported by Salazar, 2026). In its September 2026 inflation report, Peru's central bank projected a 30.7% contraction of the fishing sector for 2026. Research commissioned by Peru's National Fisheries Society puts about 78,000 fishing-sector jobs at risk in 2026 and another 44,000 in 2027 (Salazar, 2026). Peru's anchovy lives in a different current system (the Humboldt), but the mechanism is the same one this project studies: warm water suppressing the upwelling that small fishes depend on.

The current West Coast event carries similar risks. NOAA Fisheries and partners are tracking NEP25A because marine heatwaves can trigger harmful algal blooms that sicken marine mammals and close shellfish fisheries, and past heatwaves, including the Blob, have been linked to lower ocean survival of salmon (Responsible Seafood Advocate, 2026). An early harmful algal bloom off Southern California in 2025 killed hundreds of sea lions, dolphins, and seabirds, and large numbers of tuna were caught in Alaska in autumn 2025, an area where they are not normally found in such numbers (Responsible Seafood Advocate, 2026).

**Why we look at fish spawning (RQ3).** A heatwave map only helps fisheries if it connects to the species they depend on. Pacific sardine and northern anchovy are the main forage fish of the California Current: prey for larger fish, marine mammals and seabirds, and the basis of commercial fisheries. Both reacted strongly to the Blob, in opposite directions. Sardine collapsed before and during the heatwave, leading to a multi-year federal fishery disaster and closure from California to Washington, while anchovy began increasing during the heatwave and reached near-record levels afterwards (Free et al., 2023). Sardine shifted its distribution markedly; anchovy largely stayed in its usual spawning areas (Muhling et al., 2020).

These opposite responses raise the question RQ3 asks: was the sardine shift linked to the warm water itself, and could heatwave exposure near the coast flag where spawning is likely to move? Answering it needs fish observations that overlap the heatwaves in time and place, which our tutor asked for at the poster stage. Spawning is also where temperature matters most for future catches, because adults choose where to release eggs, and egg and larval survival sets the size of later year classes (Robidas et al., 2026). A review of the Blob's effects on West Coast fisheries recommends expanding monitoring to provide early warning signals of impacts (Free et al., 2023). A drifter-based measure of heatwave exposure on spawning grounds is one such signal.

**Why two similar events had different outcomes.** The 2019–2020 heatwave was nearly as large as the Blob but stayed far offshore, while the strong upwelling kept the northern California Current cool and productive (Weber et al., 2021). Size alone therefore does not predict impact. What matters is where the warm water goes and what it reaches.

NOAA scientists make the same point about the current event. Andrew Leising, who runs the Blobtracker, notes that NEP25A has rivalled the Blob in area but so far had much less ecological impact, because it has not reached as deep into the water column and has not spent as much time near the coast (Responsible Seafood Advocate, 2026). Time near the coast is exactly what drifters can measure.

**Who can use this study:**
- **NOAA and the GDP:** the project tests a new use of an existing observing network, as a low-cost, real-time check on heatwave conditions where drifters are.
- **Fisheries managers:** it asks whether heatwave exposure near the coast is linked to where commercially important fish spawn.
- **Ocean science:** it compares Lagrangian and Eulerian views of the same events.

 </p>

<font color="#0077B6" size="5px">**4. Literature Review and Existing Approaches**</font>
---
<p align="justify"> Some existing work gave a standard heatwave definition, an operational tracker, a validated drifter dataset and detailed accounts of both events; however, none of it combines drifter movement, heatwave detection and fish observations.

**Defining marine heatwaves.** Hobday et al. (2016) define a marine heatwave as a period of at least five days with SST above the 90th percentile of a 30-year climatology for that location and day of year. Because the threshold varies by place and season, the definition works in both warm southern and cold northern water. It is the standard in the field, and we use it for every research question. NOAA's Blobtracker uses a related area-based rule: a large heatwave is a contiguous region where the normalised SST anomaly exceeds 1.29 standard deviations, covering more than 400,000 km² (NOAA California Current IEA, 2026). The two definitions answer different questions, one per grid cell and one per event, so we compare their timing and location rather than expecting identical flags.

**Satellite SST.** NOAA's Daily Optimum Interpolation SST (OISST) v2.1 blends satellite, ship and buoy measurements, including drifters, onto a daily 0.25° grid from 1982 (Huang et al., 2021). It is the reference for both the Blobtracker and our analysis. Because it already assimilates drifter SST, agreement between drifters and OISST is a consistency check, not fully independent validation.

**Drifter data.** Elipot et al. (2016) describe the GDP hourly product, including how hourly positions and velocities are derived. Elipot et al. (2022) document the hourly SST product and its quality flags. The clouddrift Python package gives cloud access to these data as "ragged arrays" (Elipot et al., 2024). Drifters can lose their drogue (sea anchor); after that they follow the wind as well as the current, which matters for movement analysis.

**The two events.** Elipot et al. (2016) describe the GDP hourly product, including how hourly positions and velocities are derived. Elipot et al. (2022) document the hourly SST product and its quality flags. The clouddrift Python package gives cloud access to these data as "ragged arrays" (Elipot et al., 2024). Drifters can lose their drogue (sea anchor); after that they follow the wind as well as the current, which matters for movement analysis.

**Fish and heatwaves.** CalCOFI's Continuous Underway Fish Egg Sampler (CUFES) pumps surface water and counts sardine and anchovy eggs while a ship is moving (Checkley et al., 1997). Studies of spawning temperatures report that anchovy spawning is concentrated in 11.5–16.5 °C (Lluch-Belda et al., 1991), and that most sardine eggs off the U.S. and Mexico were found below 17 °C (Valencia-Gasti et al., 2018). During the Blob, sardine distributions shifted markedly while anchovy largely kept their historical spawning areas, and species distribution models trained on earlier years lost skill under the new conditions (Muhling et al., 2020). The post-Blob anchovy boom was unexpected because anchovy were thought to favour cold water; recent work links it to larval condition and food-chain efficiency rather than temperature (Swalethorp et al., 2023; Robidas et al., 2026).

 </p>

<font color="#0077B6" size="5px">**5. Research Gaps**</font>
---
<p align="justify"> Existing marine heatwave monitoring, such as NOAA’s Blobtracker, focuses on where warming and unusual changes in water temperature occur, but does not fully capture how ocean circulation and upwelling influence the transport of warm water, where it moves and which ecosystems or fisheries are most exposed (NOAA Southwest Fisheries Science Center, n.d.). By incorporating drifter trajectories, velocity and temperature, we aim to distinguish which heatwaves have greater potential to affect coastal ecosystems and fisheries. Paired with fish egg data, these observations could also serve as an early-warning tool for fisheries at risk from marine heatwaves, particularly given the potential for intensified ocean warming associated with the forecast 2026-27 El Niño event.

<p align="justify"> <strong>Drifter Cleaning.</strong> For temperature we kept only observations with the highest SST quality flag (flg_sst = 5). For movement we kept observations with both eastward and northward velocity (98.6% complete in 2019–2021). Hourly records were averaged into one value per drifter per day where a day had enough hourly readings, giving daily "drifter-days" for matching to OISST.

<strong>Fish data.</strong> CUFES samples eggs at about 3 m depth every 30 minutes while a ship is underway (Checkley et al., 1997). Eggs mark where adult fish chose to spawn, so they are a direct record of where fish were, independent of fishing effort. The sardine fishery was closed from 2015, which makes catch data unsuitable for this period.

<strong>7.2 Aligning the datasets in space and time</strong>
OISST is the common grid. Every drifter-day and every egg sample is matched to the nearest OISST grid cell (0.25°) on the same day. For each match we record the OISST 90th-percentile threshold for that cell and day of year, whether the cell was in a heatwave, and, for egg samples, the share of the previous 30 days the cell was in a heatwave. To check that the temperatures agree with what fish experienced, we also paired drifter-days with ship SST at egg samples within 50 km and one day.

<strong>7.3 Heatwave definition (all research questions)</strong>
Following Hobday et al. (2016), we computed for each OISST cell and day of year the 90th percentile of SST over a 30-year baseline, using an 11-day window centred on that day and smoothing with a 31-day moving average. A cell is in a heatwave when SST stays above this threshold for at least five consecutive days. The RQ3 analysis uses 1982-2011, which ends before both events so neither inflates its own threshold. The RQ1 and RQ2 analyses used 1983-2012 (Section 10).

<strong>7.4 Research Question 1</strong>

To determine whether the drifters could consistently track both heatwave events accurately, two marine heatwave detection models using the GDP drifter SST observations were developed, with their performance compared against the OISST marine heatwave reference and NOAA’s Blobtracker.

Model A serves as the simple baseline detection model, which uses GDP drifter SST observations to identify unusually warm conditions. It calculates temperature anomalies by comparing observed GDP SST with the historical mean SST for the corresponding time and location of that year. If the calculated anomaly is above +1°C, model A flags the observation as above normal temperature and therefore as anomalously warm.

Model B uses a statistical threshold, identifying GDP observations as anomalously warm if measured SST exceeds the historical 90th-percentile threshold for the corresponding time and location of that year. This threshold is calculated using historical OISST climatology, making model B able to account for seasonal and regional variations in temperature. However, as Model B uses the same historical OISST 90th-percentile threshold as the reference dataset, its detection method is more closely aligned with the evaluation criteria. This should be considered when interpreting differences in model performance.

The performance of both models was evaluated by comparing their marine heatwave detections, based on GDP sea surface temperature (SST) anomalies, against an OISST reference dataset. In the OISST dataset, marine heatwaves were defined as periods when SST exceeded the historical 90th percentile for at least five consecutive days. Each GDP observation was matched to the OISST reference using the corresponding date and nearest spatial grid cell. This enabled an assessment of how accurately each model identified marine heatwaves relative to the OISST reference. To evaluate which model demonstrated stronger agreement with the OISST reference data, models precision and F1 scores were compared.

<strong>Hypothesis:</strong>
Model B is hypothesised to achieve greater accuracy in detecting marine heatwave conditions, as its historical 90th-percentile threshold accounts for seasonal and regional variations in SST. In contrast, Model A applies a fixed +1°C anomaly threshold, which does not account for spatial differences in temperature variability. Consequently, Model B is expected to demonstrate stronger agreement with OISST-defined marine heatwave conditions.  

<strong>Sensitivity Analysis of Marine Heatwave Detection Thresholds</strong>

However, as ocean temperatures continue to rise due to climate change, fixed historical thresholds do not account for changes in background SST conditions over time, potentially affecting the models' ability to identify anomalously warm conditions during future events, such as a potential 2027 El Niño. To investigate this limitation, further analysis was conducted to assess whether adjusting for historical SST trends influenced the frequency of anomalously warm observations detected by Models A and B during the 2013–2016 Blob and 2019–2020 marine heatwave events. Additionally, threshold tuning and temporal generalisation assessments were conducted to determine whether alternative detection thresholds could improve model performance and whether these improvements remained consistent across both marine heatwave events.

<strong>7.5 Research Question 2</strong>

To examine differences in warm-water movement and coastal exposure between the Blob and the 2019–2020 marine heatwave, GDP drifter trajectories and the coastal proximity of anomalously warm water were compared across both events. Anomalously warm conditions were identified by comparing GDP SST observations against the historical OISST 90th-percentile threshold and the marine heatwave reference established in RQ1, which required temperatures to exceed this threshold for at least five consecutive days. Coastal proximity was then determined by calculating the distance between each drifter observation and the mapped coastline, with observations located within 50 km classified as coastal. Applying this consistent geographical threshold enabled a standardised comparison of warm-water distribution and coastal exposure between the two marine heatwave events.

<strong>Hypothesis.</strong>

Based on previous research, the 2013-2016 Blob extended into coastal waters and affected coastal ecosystems, whereas the 2019-2020 marine heatwave remained largely offshore (Weber et al., 2021). Therefore, anomalously warm GDP drifters were expected to be observed closer to the coastline and to record more entries during the Blob. These differences would indicate greater coastal exposure among the sampled drifters during the earlier event.

<strong>Observed drifter movement and coastal exposure.</strong>

Observed warm-condition GDP drifter movements were compared across the two marine heatwaves. Coastal proximity was assessed by calculating the median distance to the mapped coastline for each drifter and comparing the distributions between events. A coastal entry was defined as an observed transition from outside to within the 50 km coastal zone. Crossing frequencies were expressed per 100 offshore drifter-days to account for differences in offshore observation exposure between events. The 24-hour net displacement was calculated as the geographical distance between a drifter’s starting and observed endpoint positions, rather than the total path length. These measures distinguished proximity to the coast, entry into the coastal zone and the extent of movement.

<strong>Trajectory Prediction Models.</strong>

To glean insights into the movements of drifters during the two heatwaves, three trajectory prediction models were evaluated to estimate the geographical positions of GDP drifters over a 24-hour forecast period. All three models used the same starting positions and prediction horizon, allowing their performance to be compared under consistent conditions.

Model A (Persistence) served as the baseline model. It assumed that a drifter remained stationary over the following 24 hours, meaning its predicted endpoint was identical to its starting position.

Model B (Constant Velocity) estimated the future position using the drifter's current eastward and northward velocities. It assumed that the observed velocity remained constant throughout the 24-hour prediction period.

Model C (Historical Velocity) used the average eastward (ve) and northward (vn) velocities from the previous six hourly observations. Unlike Model B, which relied on the current velocity, Model C incorporated recent velocity measurements to estimate the drifter's future position. The comparison tested whether velocity information improved predictions relative to the stationary baseline, and whether averaging recent velocities improved on a single velocity estimate. Models B and C held their respective velocity estimates constant over the forecast period, and neither model represented changes in currents during the following 24 hours.

<strong>Model evaluation.</strong>

The three trajectory models were evaluated across two metrics; positional accuracy in kilometres by comparing their predicted 24-hour endpoints with the actual GDP positions observed after 24 hours, and coastal exposure classification accuracy. Predicted and observed 24-hour endpoints were classified as "coastal" if they fell within 50 km of the mapped coastline. Precision (the proportion of predicted coastal endpoints that were actually coastal), recall (the proportion of observed coastal endpoints that were correct predicted by the model) and F1 scores (how well the model balanced precision and recall) were used to measure classification accuracy rather than overall accuracy because offshore observations heavily outnumbered coastal endpoints, creating a class imbalance. This metric assessed endpoint presence within the 50-km coastal zone rather than isolating new coastal entries.

<strong>Temperature Persistence.</strong>

A temperature-persistence baseline was also evaluated to assess whether unusually warm conditions persisted at drifters' later observed locations, as trajectory models only predicted drifter movement. The baseline assumed that each drifter's current warm or non-warm classification would remain unchanged over time. These predictions were then compared with observed SST classifications after 24, 48 and 72 hours, with performance assessed using precision, recall and F1 scores. Comparing performance at 24, 48 and 72 hours assessed how the baseline’s reliability changed with forecast lead time. This tested the persistence of a temperature category rather than predicting a numerical SST value.

<strong>7.6 Research Question 3: Exposure of spawning habitat and egg models</strong>
<strong>Species ranges.</strong> A working preferred range of 11.5–16.5 °C was used for northern anchovy, the spawning range reported by Lluch-Belda et al. (1991), and 12–17 °C for Pacific sardine, whose upper limit matches the 17 °C below which most sardine eggs were found by Valencia-Gasti et al. (2018). Published sardine ranges differ between studies and stocks (Lluch-Belda et al., 1991, report 13–25 °C across the whole system), so ranges shifted by ±2 °C were also tested.

<strong>Fair comparisonship.</strong> Ship surveys cover different areas in different years, so exposure was compared only for spring samples (March to May) in 31.5–35.5°N, an area surveyed every year. The periods were Before (2012–13), Blob (2014–16), Between (2017–18), 2019–20 and After (2021).

<strong>Egg models.</strong> Each model predicts whether a sample contains eggs (present or absent), with three nested input levels:

- <strong>A, place and season:</strong> latitude, longitude and day of year.
- <strong>B, plus temperature range:</strong> whether the water is inside the species’ range, and how far above or below it.
- <strong>C, plus heatwave exposure:</strong> whether the sample is in a heatwave, the share of the previous 30 days in a heatwave, and the share above the species’ upper limit.

If C beats A, heatwave exposure adds information beyond where and when the species usually spawns. Three model sets used these same inputs, chosen to range from simple to flexible so the result does not depend on one model type:

- 1. <strong>Rule-based overlay (no fitting):</strong> the usual egg rate for that 1° latitude band and month in the other years, halved when the water is outside the range and halved again after recent heatwave exposure. This is the overlay idea from the project proposal.
- 2. <strong>Logistic regression:</strong> adds up the evidence (usual place, usual season, recent heatwave) and turns it into a chance of finding eggs. It is simple and easy to explain.
- 3. <strong>Gradient boosting (scikit-learn HistGradientBoostingClassifier; Pedregosa et al., 2011):</strong> many small decision trees that can capture combinations the simple model cannot, such as heatwave exposure mattering in the south in spring but not in the north.

<strong>Evaluation and decision rule.</strong> Every model was tested by leave-one-year-out cross-validation: each year is predicted by a model trained on the other years, so the model cannot memorise the Blob. The main metric is ROC-AUC, the chance that a sample with eggs is ranked above one without (0.5 means no skill, 1 means perfect). PR-AUC and the Brier score were also checked. The chosen set is the simplest one (rule, then logistic, then boosting) whose best model is not clearly worse than the best model overall, meaning the 95% interval of the difference includes zero.

<strong>Sparse data handling and overfitting.</strong> Three kinds of sparse data affect RQ3:
- <strong>Sparse drifters.</strong> Drifters cover only some places, so exposure comes from the gap-free OISST grid, and drifters are used as a check (Section 8.3).
- <strong>Uneven ship surveys.</strong> Surveys do not cover the same area every year, so exposure is compared only within the fixed spring area, and the sardine latitude shift compares only the springs when northern lines were surveyed (2012, 2015 and 2016).
- <strong>Patchy eggs.</strong> Most samples contain no eggs, so the models predict presence rather than counts and are scored with AUC, which is not inflated by the many zeros. Uncertainty intervals come from a bootstrap that resamples whole survey days, because samples from the same day are near-copies; resampling single samples would make the intervals look falsely narrow.

To stop gradient boosting from overfitting, the trees were kept small (3 levels deep, at least 40 samples per leaf, learning rate 0.05, L2 regularisation 1.0), every score is on a year the model never saw, and a result is only claimed when the simpler sets agree.

<p align="justify"> <strong>8.1 RQ1: Can drifters detect the heatwaves?

Investigating how effectively drifters detect the Blob and the 2019-2020 marine heatwave (Methodology 1)

Which model performed better? </strong>

The confusion matrices compare each model's predictions against the OISST reference, and were used to calculate precision, recall and F1 scores summarised in Figure 1.2. Model B correctly identified more marine heatwave observations than Model A (322,966 vs 318,402), reducing false positives by 34.7% and false negatives by 23.9%. Figure 1.2 shows Model B achieving higher precision (0.802 vs 0.723), recall (0.957 vs 0.944) and F1 score (0.873 vs 0.819). Moreover, figure 1.3 shows that Model B maintained consistently higher F1 scores across both events (0.874 for the Blob and 0.871 for the 2019-20 MHW), demonstrating stronger and more consistent agreement with the OISST reference.

To assess how effectively Models A and B captured the timing of marine heatwave events, their detection rates were compared with the OISST reference, revealing broadly similar temporal patterns. <strong>Can the drifters track the spatial distribution of the Blob and the 2019-2020 marine heatwave?</strong>

The spatial distribution maps in Figure 1.5 assess how accurately GDP drifters capture marine heatwave conditions relative to the OISST reference data. In October 2014, OISST indicated widespread warming, while September 2019 showed stronger offshore warming and weaker coastal warming. As shown in Figure 1.6, these patterns broadly aligned with NOAA's BlobTracker observations, although some differences in cooler offshore regions during October 2014 were different. This suggests that GDP observations are better suited for identifying localised warming rather than mapping entire marine heatwave events.

<strong>Methodology 1 results:

Test 1: Was a historical SST trend observed in the study region during 1983-2012?</strong>

image.png

<em>Figure 1.7. Annual mean SST and estimated linear trend in California Current (1982-2012)</em>

Overall, a historical SST trend of -0.170°C per decade was estimated, indicating an overall cooling trend during the baseline period. It was found that this negative trend resulted in progressively lower temperature thresholds for later years, potentially increasing the number of GDP observations classified as anomalously warm. However, further statistical validation is needed to confirm this trend and assess whether fixed historical thresholds are suitable for studying later marine heatwaves.

<strong>Model test sensitivity to different adjustment strengths</strong>

image.png

<em>Figure 1.8. Model B’s sensitivity test against a fixed OISST reference</em>

image.png

<em>Figure 1.9. Model B’s sensitivity test against an adjusted OISST reference </em>

The sensitivity test examined how historical trend adjustments influence Model B’s detections. Increasing the adjustment from 0% to 150% raised the proportion of anomalously warm GDP observations from 55% to 72% during The Blob and 38% to 61% during the 2019–2020 marine heatwave. This occurs because greater detrending lowers the P90 threshold, making SST values more likely to exceed it. Both Model B and the trend-adjusted OISST reference detected more heatwave conditions as the adjustment increased, while the original fixed OISST reference remained unchanged. Overall, these findings demonstrate that Model B’s detections are sensitive to the historical baseline used.

image.png

<em>Figure 1.10. Model B’s detection performance at a fixed OISST vs Adjusted OISST for The Blob and 2019-2020 marine heatwave</em>

When OISST remained fixed, increasing trend adjustment reduced F1 scores from approximately 0.87 to 0.78 during The Blob and 0.87 to 0.69 during the 2019 marine heatwave, indicating weaker agreement with the original reference. Conversely, when both GDP and OISST thresholds were adjusted, F1 scores increased to approximately 0.93 and 0.92, respectively, indicating greater agreement of anomaly classification when datasets both use progressively lower P90 thresholds. Overall, these findings reinforce the findings from figure 1.7, demonstrating that Model B’s measured performance depends on the baseline and parameters set. While consistent adjustments improve agreement, higher F1 scores do not necessarily indicate greater accuracy or prove that detrending is superior to the original marine heatwave definition.

<strong>Robustness and sensitivity analysis results (Methodology 2):</strong>

<em>Figure 1.11. Robustness check of Model B’s detection performance across different GDP thresholds against the fixed OISST P90 five-day reference</em>

<strong>Is the P90 threshold the most suited threshold for marine heatwave detection?
</strong>

Model B’s 90th percentile threshold was initially chosen to stay consistent with the OISST marine heatwave reference definition. Figure 1.11 supports this decision, as P90 achieved the highest F1 scores across both marine heatwave events. However, P95 performed equally well during the 2013-2016 Blob, showing the threshold's effectiveness may vary across datasets, regions and marine heatwave events.

<strong>Test 2: Sensitivity Test for Consecutive Day-Duration</strong>

image.png

<em>Figure 1.12. Duration robustness check for Model A and Model B</em>

<strong>Does changing the heatwave definition drastically affect results?</strong> Following Hobday et al. (2016), marine heatwaves for this project were defined as periods when SST exceeds the historical 90th-percentile threshold for at least five consecutive days. Figure 1.12 shows that F1 scores declined slightly as the minimum duration increased, but Model B consistently outperformed Model A across both marine heatwave events. This indicates that although duration requirement affected F1 scores, it did not change the relative performance of the two models. Therefore, the five-day threshold is a valid and consistent measure with the standard marine heatwave definition.

<strong>Test 3: Threshold Duration Grid</strong>

image.png

<em>Figure 1.13. Threshold-duration grid to check how changing both GDP temperature threshold and minimum duration requirement affects detection </em>

The P90 threshold consistently achieved the highest F1 scores across both marine heatwave events. During The Blob, three-day and five-day durations achieved the highest F1 score of 0.90, whilst one-day and three-day durations performed best during the 2019-2020 marine heatwave (0.89). These findings suggest that longer duration requirements do not necessarily improve detection performance. However, given the minimal F1 difference of 0.01, the model continues to use the five-day requirement established by Hobday et al. (2016) to maintain consistency with the standard marine heatwave definition. Hence, Model B’s chosen parameters of the 90th percentile threshold and five-consecutive-day duration can be considered robust and appropriate, as they maintained performance across the tested variations, with only minor changes in accuracy.

<strong>Test 4: Threshold Tuning and Temporal Generalisation</strong>

image.png

<em>Figure 1.14. Effect of SST Anomaly Threshold on Model A’s F1 Score during the 2013-2016 Blob</em>

Seven fixed anomaly threshold <em>k{0.5, 0.75, 1.0, 1.25, 1.5, 1.75, 2.0} degrees C</em> above the historical OISST mean for Model A was tuned, and the threshold with the highest F1 score (+1.25°C, with 0.8410) was compared against the OISST P90 +1°C reference. When applied to the holdout 2019-2020 marine heatwave event without modification, the tuned +1.25°C detector achieved a reduced F1 of 0.7988, suggesting a reduction in agreement with the OISST reference compared with the earlier event. In the end, model B still maintained higher F1 scores across both events. However, this was not a completely independent holdout test, so further testing with other events is needed to confirm whether the selected threshold generalises reliably to unseen marine heatwave conditions.

<strong>Coastal endpoint classification performance</strong>

<em>Figure 2.4. 24-hour coastal endpoint classification performance by model and marine heatwave event</em>

Coastal-endpoint classification assessed whether predicted drifter positions were within 50 km of the mapped coastline after 24 hours. Since most observations were offshore, overall accuracy could be misleading. Instead, three models were evaluated using precision, recall and F1 scores. Model C achieved the highest F1 scores across both events (0.920 for the Blob and 0.538 for the 2019–2020 marine heatwave), with recall scores of 0.926 and 0.817, respectively. Performance was weaker during the 2019-2020 marine heatwave event, where Model C correctly identified 67 of the 82 coastal endpoints but produced 100 false positives, resulting in a low precision of 0.401. <strong>Temperature persistence</strong>

At 24 hours, the warm-class F1 score was 0.943 for the Blob and 0.916 for the 2019–2020 marine heatwave. Overall, sampled warm-condition drifters were generally closer to shore during the Blob than during the 2019–2020 marine heatwave. The temperature-persistence baseline also performed well over short horizons, although F1 scores declined as the forecast period increased. Uneven drifter coverage and the limited number of coastal endpoints during 2019–2020 restrict the generalisability of these findings.

<p align="justify"> <strong>8.3 RQ3: Did the exposure reach the fish?</strong>

<strong>Are the temperatures right?</strong> Drifter SST matched ship SST at egg samples closely: 663 pairs within 50 km and one day, mean difference −0.08 °C, correlation 0.93. OISST matched ship SST with no average bias across all 17,269 samples (standard deviation 0.82 °C). The temperatures used are therefore the ones the eggs were sampled in, although OISST already includes ship and drifter data, so this is a consistency check rather than independent proof.

<strong>Exposure of spawning habitat.</strong> During the Blob, 44% [33–55] of spring samples in the core area were in a heatwave, against 10% [3–18] in 2019–20 and 1% before the Blob (Figure 3.1, Figure 3.2). The intervals do not overlap: the Blob exposed spawning habitat about four times as often, which agrees with RQ2. Yet even at its peak, only 8% of samples were above the sardine upper limit (17 °C) and 15% above the anchovy limit (16.5 °C). The Blob made habitat unusually warm for the place and season far more often than it made it too hot for the fish. Drifters in the same area and season showed the same contrast (59% vs 19% of drifter-days in a heatwave), from 25 and 6 drifters.

image.png

<em>Figure 3.1. Share of spring egg-survey samples (31.5–35.5°N) exposed, with 95% intervals from a bootstrap over survey days. 2021 is left out (10 survey days).</em>

image.png

<em>Figure 3.2. Heat exposure of spawning habitat by period: share of spring samples in a heatwave and above each species’ upper limit, with 95% intervals.</em>

<strong>Where fish spawned.</strong> Sardine eggs nearly disappeared from the core area during the Blob, falling from 26% [19–33] to 2% [1–4] of samples, and the median  latitude of sardine eggs moved from 34.9°N to 42.6°N, about 850 km north, off Oregon (Figure 3.3). The water temperature where sardine eggs were found barely changed (12.9 to 12.8 °C): sardines appear to have moved with their preferred water rather than putting up with warmer water. Anchovy did the opposite. Their egg presence rose from 0% before to 5% during, 29% between and 36% in 2019–20, without moving north, matching the known anchovy boom after the Blob (Thompson et al., 2022; Free et al., 2023). The northern survey lines (38°N and above) were only sampled in spring 2012, 2015 and 2016, so the latitude shift compares exactly those years and rests on a small 2012 baseline.

image.png

<em>Figure 3.3. Spring sardine egg samples in 2012 (before the Blob) and 2015–16 (Blob peak); grey points were surveyed with no sardine eggs. Right: yearly share of spring samples with eggs in the 31.5–35.5°N core, with median ship SST.</em>

<strong>Is exposure linked to where eggs occur?</strong> Maps can mislead, so the link was tested with nine models per species (three sets at three input levels), each predicting years it had not seen (Figure 3.4, Figure 3.5). For sardine, adding heatwave exposure (C − A) improved prediction in every set, with every interval above zero: +0.015 for the rule-based ovrlay, +0.084 for logistic regression and +0.029 for gradient boosting. Within the logistic set, each step added skill: AUC 0.584 (A), 0.633 (B), 0.669 (C). For anchovy, the gain was clear only in the logistic set (+0.015) and was about zero or negative in the others.

image.png

<em>Figure 3.4. Gain in leave-one-year-out ROC-AUC from adding heatwave exposure within each model set, with 95% bootstrap intervals, and the set chosen by the decision rule.</em>

image.png

<em>Figure 3.5. Leave-one-year-out ROC-AUC of every egg model, grouped by model set, for sardine and anchovy (95% intervals). Text shows the gain from adding heatwave exposure (C − A); an asterisk marks an interval above zero.</em>

<strong>Choosing the best set.</strong> For sardine, gradient boosting is clearly better than the logistic (0.669) and rule-based (0.628) sets, so the decision rule accepts its extra complexity: sardine’s response to temperature depends on latitude and season together, which trees can capture. For anchovy, logistic regression (0.789) is the simplest set not clearly worse than boosting (0.801), so it is chosen. The rule-based overlay, the original proposal idea, was weakest for both species (0.63 and 0.70), so an overlay alone would have under-used the data.

<strong>Robustness.</strong> The sardine gain stayed positive when the species ranges were shifted by −2 to +2 °C (+0.035 to +0.085) and for exposure windows of 10, 30 and 60 days (+0.062 to +0.099), and was largest near the published range. Anchovy gains stayed small and turned slightly negative at −1 °C.

<strong>Limits of the models.</strong> No model reproduces the size of the sardine collapse or the anchovy boom, and predicted probabilities were no better calibrated with exposure (Brier score unchanged). The models capture where eggs are likely within a survey, not how many fish there are. An AUC of 0.72 is moderate, but fish eggs are very patchy, and the gain over place and season is consistent across model types.

<strong>RQ3 answer.</strong> The Blob exposed spawning habitat to heatwave conditions about four times as often as 2019–20, although most water stayed inside the species’ ranges. Sardine spawning moved about 8° north into water of the same temperature, and heatwave exposure improved predictions of where sardine eggs occur in every model set and under every range and window tested. For anchovy, exposure did not reliably help, which agrees with evidence that the anchovy boom was driven by larval survival and food rather than temperature (Swalethorp et al., 2023; Robidas et al., 2026). This is a link, not proof of cause.


</p>

<font color="#0077B6" size="5px">**9. Result and Findings**</font>
---
<p align="justify"> Drifters detect heatwave conditions locally, the two events differed mainly in coastal reach, and that reach lined up with a northward shift in sardine spawning.

image.png

## Recommendation

NOAA and the GDP should use drifter observations as a <strong>local, real-time check alongside the Blobtracker</strong>, not as a replacement:

- <strong>Flag with the 90th-percentile rule.</strong> Flag drifter observations above the OISST 90th percentile using the standard five-day definition. It matched OISST best and carried over best to an unseen event; tuned alternatives overfit.
- <strong>Watch coastal reach, not just size.</strong> For a detected heatwave, track the share of warm-condition drifters near the coast and their coastal-entry rate. This, not area, separated the damaging Blob from the 2019–20 event, and NOAA explains NEP25A’s smaller impact so far in the same way.
- <strong>Treat shoreward warm water as an early signal for sardine.</strong> When warm water is heading towards the shore in spring, sardine spawning grounds off southern and central California are the first place to look for a northward shift. For 2026–27, with NEP25A growing off the southern West Coast and a strong El Niño likely, this is a monitoring priority, not a forecast of fish numbers.
- <strong>Do not rely on temperature alone for anchovy.</strong> Their response was not explained by heat exposure.

</p>

<font color="#0077B6" size="5px">**10. Limitations**</font>
---
<p align="justify">Whilst ambitious in scope, the project does have several limitations affecting the accuracy of our results, <strong>causing constraints on the depth of our analysis.</strong> The project aimed to forecast for the 2027 El Niño, however the GDP release ends in October 2022, and does not provide <strong>recent</strong> conditions during 2025-26. Comparing the data for accuracy became difficult as our reference (OISST) had missing days from 1992-98. This study primarily aimed to explore how warm water moves in coastal areas, which became difficult as drifters sample unevenly, which became challenging as little coastal data from 2019-20 was found.

The historical analysis may inform discussion of future monitoring applications, but it should not be interpreted as direct evidence of ecological conditions or impacts during 2025-26. Velocity-data limitation: Although velocity coverage is high, drifter velocity may also depend on drogue status. The drogue acts as a sea anchor and allows the buoy to more closely follow subsurface water motion; after drogue loss, recorded velocity can be more affected by wind slip. The current preliminary analysis does not explicitly separate observations by drogue status, so this potential source of bias should be considered when interpreting transport and coastal-exposure results. Comparing the Blob and the 2019-20 Marine heatwave to relevant fish egg data was challenging due to difficulties aligning datasets spatially and temporally.  Fish egg surveys were limited for 2019-2020 and only reflected spawning activity rather than overall fish abundance. Additionally, pre-existing declines in sardine populations made it difficult to isolate the impacts of marine heatwaves on fisheries.
</p>

<font color="#0077B6" size="5px">**11. Future Work**</font>
---
<p align="justify">Future research could extend this study by predicting marine heatwave patterns and hotspots associated with the anticipated 2026-2027 El Niño, once more recent drifter data from NEP25A and NEP26A becomes available. Building on RQ2, the trajectory models could be extended beyond 24 hours and evaluated against later GDP positions. Temperature-persistence evaluation could also be extended beyond the 72-hour horizon already tested. Additionally, future studies could incorporate fish population and catch data, rather than relying solely on fish eggs, provided sufficient spatially and temporally aligned datasets become available. This approach could also be expanded to investigate the impacts of marine heatwaves on other declining marine species and flora.</p>

<p align="justify"> AI tools (ChatGPT and Claude) were used to search for literature and datasets, debug code, plan model comparisons, and draft and check text. All numbers in this report come from the project code, and every reference was opened and checked against its source.

<strong>Where AI helped:</strong>

- <strong>Finding the fish data.</strong> An AI-assisted search pointed to NOAA's CalCOFI CUFES egg data, which became the independent fish dataset for RQ3.
- <strong>Fixing the OISST download.</strong> The first loader silently saved empty files for 1982–1990 and gappy files for 1992–1998, which would have distorted the 90th-percentile baseline. AI help traced this to the download request and replaced it with direct ERDDAP downloads plus a check that rejects any year with too few days (Section 7.7).
- <strong>Planning model comparisons.</strong> AI suggested comparing whole model sets (rule-based, logistic regression, gradient boosting) with nested inputs and a stated decision rule, which shaped the RQ3 design and the RQ1 event-level forward test.

<strong>Where AI was wrong, and how it was caught:</strong>

- <strong>A wrongly attributed temperature range.</strong> An AI-drafted note credited a sardine range of 12–17 °C to Valencia-Gasti et al. (2018). The paper itself reports that most eggs were found below 17 °C and at average SSTs of 13.0–14.3 °C in US waters. The range was kept as a working choice, described accurately in Section 7.6, and tested with ±2 °C shifts.
- <strong>Changing the research question.</strong> At one point AI recommended replacing RQ3 with a question about detection timing. That moved away from the proposal and its motivation, so the species-exposure question was kept and the timing work became supporting analysis for RQ1.
- <strong>Outdated and garbled facts.</strong> AI-assisted drafts carried an older “greater than 90%” El Niño probability, which was updated to NOAA's latest outlook. A citation generator also turned the GDP author “Centurioni, L.” into “Luca, C.”, which was corrected.


Overall, AI was most useful for finding data and debugging, and least reliable for facts and citations, which is why every number was traced back to code output and every reference to its source.
